```python
# Household Monthly Budget Analyzer

print("HOUSEHOLD MONTHLY BUDGET ANALYZER")
print("----------------------------------")

# Taking input from user
income = int(input("Enter your total monthly income: "))

rent = int(input("Enter rent expense: "))
food = int(input("Enter food expense: "))
transport = int(input("Enter transport expense: "))
utilities = int(input("Enter utilities expense: "))
other = int(input("Enter other expenses: "))

# Calculating total expenses
total_expense = rent + food + transport + utilities + other

# Calculating savings or deficit
difference = income - total_expense

print("\n----- Monthly Summary -----")
print("Total Income:", income)
print("Total Expenses:", total_expense)

if difference > 0:
    print("You have a SURPLUS of:", difference)
    print("Good job! You are saving money.")
elif difference < 0:
    print("You have a DEFICIT of:", abs(difference))
    print("Warning! You are spending more than your income.")
else:
    print("Your income and expenses are equal. No savings, no deficit.")

print("---------------------------")
```
