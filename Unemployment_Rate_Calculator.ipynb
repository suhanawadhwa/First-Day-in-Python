```python
# Unemployment Rate Calculator

print("UNEMPLOYMENT RATE CALCULATOR")
print("----------------------------")


total_labor_force = int(input("Enter total labor force: "))
unemployed_people = int(input("Enter number of unemployed people: "))


unemployment_rate = (unemployed_people / total_labor_force) * 100

print("RESULT")
print("Total Labor Force:", total_labor_force)
print("Unemployed People:", unemployed_people)
print("Unemployment Rate:", unemployment_rate, "%")


if unemployment_rate > 0:
    print("There is unemployment in the economy.")
else:
    print("No unemployment.")

```
