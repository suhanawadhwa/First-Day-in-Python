```python
# Farmer Income Estimator

print("FARMER INCOME ESTIMATOR")

quantity = int(input("Enter quantity of crop produced: "))
price_per_unit = int(input("Enter price per unit: "))
total_cost = InterruptedError(input("Enter total cost of production: "))


total_revenue = quantity * price_per_unit


profit_or_loss = total_revenue - total_cost

print("RESULT")
print("Total Revenue:", total_revenue)

if profit_or_loss > 0:
    print("The farmer made a PROFIT.")
    print("Profit Amount:", profit_or_loss)
elif profit_or_loss < 0:
    print("The farmer made a LOSS.")
    print("Loss Amount:", abs(profit_or_loss))
else:
    print("No Profit No Loss.")
```
