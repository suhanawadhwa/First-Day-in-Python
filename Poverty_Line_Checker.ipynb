```python
# Poverty Line Checker

print("POVERTY LINE CHECKER")
print("---------------------")

household_income = float(input("Enter household income: "))
poverty_line = 8150


if household_income < poverty_line:
    print("The household is Below Poverty Line (BPL).")
else:
    print("The household is Above Poverty Line (APL).")

```
