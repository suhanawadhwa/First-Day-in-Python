```python


print("BREAK-EVEN POINT CALCULATOR")

fixed_cost = int(input("Enter fixed cost: "))
variable_cost = int(input("Enter variable cost per unit: "))
selling_price = int(input("Enter selling price per unit: "))

# Break-Even Quantity = Fixed Cost / (Selling Price - Variable Cost)

break_even_quantity = fixed_cost / (selling_price - variable_cost)

print("RESULT")
print("Break-Even Quantity:", break_even_quantity, "units")

print("At this quantity, profit is zero.")
```
