```python
def start_game():
  print("Welcome to treasure game")
  print("You have to hit a crossroad")
  print("FOREST OR VILLAGE")

  choice = input("wHERE WILL YOU GO? : ").lower()
  if choice == "forest":
    forest()
  elif choice == "village":
    print("Lets go to village")
  else:
    print("Invalid input")

def forest():
  print("Welcome to forest")
  print("You have to hit a crossroad")
  print("CAVE OR RIVER")

  choice = input("wHERE WILL YOU GO? : ").lower()
  if choice == "cave":
    CAVE()
  elif choice == "river":
    print("You drowned in river")
  else:
    print("Invalid input")

def CAVE():
  print("Welcome to Cave")
  print("What are you going to do?")
  print("talk to the Old Man OR ignore the old man")

  choice = input("What will you do? : ").lower()
  if choice == "Talk to the Old Man":
    print("He gives you the map to the treasure")
  elif choice == "Ignore the old man":
    print("He casts a spell on you! booooo looser")
  else:
    print("Invalid input")

start_game()
```
