```python


print("INFLATION RATE CALCULATOR")
print("-------------------------")


old_price = float(input("Enter price of commodity last year: "))
current_price = float(input("Enter price of commodity this year: "))


inflation_rate = ((current_price - old_price) / old_price) * 100

print("RESULT")
print("Last Year Price:", old_price)
print("Current Year Price:", current_price)
print("Inflation Rate:", inflation_rate, "%")


if inflation_rate > 0:
    print("Prices have INCREASED.")
elif inflation_rate < 0:
    print("Prices have DECREASED.")
else:
    print("No change in prices.")

```
