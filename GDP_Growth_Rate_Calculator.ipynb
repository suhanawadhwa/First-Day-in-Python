```python
# GDP Growth Rate Calculator

print("GDP GROWTH RATE CALCULATOR")
print("--------------------------")


gdp_old = int(input("Enter GDP of previous year: "))
gdp_new = int(input("Enter GDP of current year: "))


growth_rate = ((gdp_new - gdp_old) / gdp_old) * 100

print("RESULT")
print("Previous Year GDP:", gdp_old)
print("Current Year GDP:", gdp_new)
print("Growth Rate:", growth_rate, "%")

# Checking economy condition
if growth_rate > 0:
    print("The economy is GROWING.")
elif growth_rate < 0:
    print("The economy is SHRINKING.")
else:
    print("No growth. Economy is stagnant.")


```
