```python


print("SIMPLE DEMAND CALCULATOR")
print("------------------------")


old_price = float(input("Enter the original price: "))
new_price = float(input("Enter the new price: "))
base_demand = float(input("Enter the base demand at original price: "))


new_demand = (old_price / new_price) * base_demand

print("RESULT")
print("Old Price:", old_price)
print("New Price:", new_price)
print("Base Demand:", base_demand)
print("New Demand:", new_demand)


if new_demand > base_demand:
    print("Demand has INCREASED because price decreased.")
elif new_demand < base_demand:
    print("Demand has DECREASED because price increased.")
else:
    print("Demand remains the SAME.")

print("------")
```
