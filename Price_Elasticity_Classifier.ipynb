```python
# Price Elasticity Classifier

print("PRICE ELASTICITY CLASSIFIER")
print("----------------------------")

# Taking input
old_price = float(input("Enter old price: "))
new_price = float(input("Enter new price: "))
old_demand = float(input("Enter old demand: "))
new_demand = float(input("Enter new demand: "))

# Calculating percentage changes
change_in_demand = (new_demand - old_demand) / old_demand
change_in_price = (new_price - old_price) / old_price

# Calculating elasticity
elasticity = change_in_demand / change_in_price

print("\nPrice Elasticity =", elasticity)

# Classifying demand
if abs(elasticity) > 1:
    print("Demand is ELASTIC")
elif abs(elasticity) < 1:
    print("Demand is INELASTIC")
else:
    print("Demand is UNITARY")


```
