```python
# Cost of Living Comparison Tool

print("COST OF LIVING COMPARISON TOOL")
print("-------------------------------")


city_a = int(input("Enter monthly expense for City A: "))
city_b = int(input("Enter monthly expense for City B: "))


if city_a > city_b:
    difference = city_a - city_b
    print("City A has higher cost of living.")
    print("Difference =", difference)

elif city_b > city_a:
    difference = city_b - city_a
    print("City B has higher cost of living.")
    print("Difference =", difference)

else:
    print("Both cities have the same cost of living.")


```
