```python
# Profit and Loss Analyzer

print("PROFIT AND LOSS ANALYZER")
print("------------------------")


total_revenue = float(input("Enter total revenue: "))
total_cost = float(input("Enter total cost: "))


result = total_revenue - total_cost

print("RESULT")

if result > 0:
    print("The business is making a PROFIT.")
    print("Profit Amount:", result)
elif result < 0:
    print("The business is making a LOSS.")
    print("Loss Amount:", abs(result))
else:
    print("No Profit No Loss. Break-even point reached.")


```
